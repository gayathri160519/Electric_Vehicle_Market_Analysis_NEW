```python
# Handle missing values # To Check
df.isnull().sum()
```

```python
# The dataset contained 10 missing values in the Legislative District column
# Missing values were found only in the Legislative District column and were handled using mode imputation.
# mode imputation is more appropriate than median imputation for Legislative District.

df['Legislative District'].fillna(
    df['Legislative District'].mode()[0],
    inplace=True
)                                                                                # o/p is just warning message not an error.
```

```python
df['Legislative District'].isnull().sum()     # After imputation, no missing values remained in the dataset.
```

```python
# Handle missing values (drop)
# Remove duplicate rows

df = df.drop_duplicates()

# Check duplicates again

print("Duplicates:", df.duplicated().sum())
df.shape
```

```python
df['DOL Vehicle ID'].duplicated().sum()       # Remove Duplicates in ID
```

```python
df['VIN (1-10)'].duplicated().sum()           # For this EV dataset, VIN (1-10) is only the first 10 characters of the full VIN, not the complete vehicle identifier.
                                              # Therefore: Duplicate VIN (1-10) values are expected. but they do not necessarily indicate duplicate records.
                                              # So i should not remove records based solely on VIN (1-10).
```

```python
df[df['VIN (1-10)'].duplicated()]            # To see the repeated VINs:
```

```python
df.rename(columns={                           # Rename column names
    'VIN (1-10)':'VIN',
    'Electric Vehicle Type':'EV_Type',
    'Electric Range':'Range_Miles'
}, inplace=True)
print(df.columns)                             # I did not drop any columns. because i need for geographical analysis.
```

```python
# Treat Outliers

Q1 = df['Range_Miles'].quantile(0.25)       # Outlier analysis was performed using a IQR method on the Electric Range column.
Q3 = df['Range_Miles'].quantile(0.75)   # these values correspond to vehicles with genuinely higher electric driving ranges and are valid observations within the EV domain.
                                            # Therefore, no outliers were removed from the dataset.
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Range_Miles'] < lower) |
              (df['Range_Miles'] > upper)]

print("Number of outliers:", len(outliers))
```

```python
# In my EV dataset, the Electric Range variable is naturally skewed.
# Many vehicles have ranges close to 0-50 miles, Some vehicles have 100-340+ miles.
# Newer EV models genuinely have much higher ranges. Because of this, IQR labels many valid high-range vehicles as "outliers".
# Removing 697 records would eliminate many legitimate EVs and distort the analysis.
```

```python
df['Range_Miles'].skew()                    # The Electric Range distribution is highly positively (right) skewed
```

```python
import numpy as np                                            # To reduce skewness and improve the distribution, a log transformation using np.log1p() was applied.

df['Electric Range Log'] = np.log1p(df['Range_Miles'])        # Final Dataset: 5000 rows, 17 columns, no missing values

print(df['Electric Range Log'].skew())                        # Distribution became more balanced.
```

```python
# Import Libraries

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")                                         # adds a white background with light grid lines to your charts.
```

```python
df = pd.read_csv('/content/EV_Cleaned_Dataset.csv')                # Load Dataset
df.head()
```

```python
# UNIVARIATE ANALYSIS                   # it is the simplest form of statistical analysis and is often the first step in exploring a dataset.
# Analyze individual columns.
```

```python
# EV Type Countplot

plt.figure(figsize=(6,4))

sns.countplot(data=df, x='EV_Type')

plt.title('Count of EV Types')
plt.xlabel('EV Type')
plt.ylabel('Count')
plt.show()
```

Interpretation of your above chart

Shows the actual number of BEVs and PHEVs.

Helps compare counts directly.



```python
# COUNTPLOT : CAFV Eligibility Distribution(Clean Alternative Fuel Vehicle)

plt.figure(figsize=(8,5))

sns.countplot(
    y='Clean Alternative Fuel Vehicle (CAFV) Eligibility', # while very few are disqualified due to low battery range.
    data=df                                                # This indicates a positive outlook for clean-energy vehicle adoption and future incentive program participation.
)

plt.title('CAFV Eligibility Distribution')
plt.xlabel('Count')
plt.ylabel('CAFV Eligibility Status')
plt.show()
```

## Interpretation of your above chart

CAFV-eligible vehicles make up a significant portion of the EV market. (Clean Alternative Fuel Vehicle).

Incentive programs play an important role in encouraging EV adoption.

Vehicles with unknown or non-eligible status highlight opportunities for further market development.

```python
# HISTOGRAM : Distribution of Electric Range

plt.figure(figsize=(8,5))
sns.histplot(df['Range_Miles'], bins=30, color='green')
plt.title('Distribution of Electric Range')
plt.xlabel('Range (Miles)')
plt.ylabel('Count')
plt.show()
```

Interpretation of your above chart

Most vehicles are concentrated within a specific range interval.

Higher-range vehicles indicate advancements in battery technology.

Manufacturers are focusing on improving driving range.



```python
# Boxplot of Electric Range

plt.figure(figsize=(8,4))
sns.boxplot(x=df['Range_Miles'])
plt.title('Boxplot of Electric Range')
plt.show()
```

Interpretation of your above chart

Outliers represent vehicles with exceptionally high electric range.

Most vehicles lie within the normal operating range.

Premium EV models may contribute to the extreme values.


```python
# PIE CHART : EV Type Distribution

ev_type = df['EV_Type'].value_counts()

plt.figure(figsize=(5,5))

plt.pie(
    ev_type.values,
    labels=ev_type.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('EV Type Distribution')
plt.show()
```

Interpretation of your above chart

Reveals whether BEVs or PHEVs dominate the market.

Shows the percentage share of each EV type.

Helps understand market composition.

```python
# BIVARIATE ANALYSIS                      # It is a statistical method used to examine the relationship between two variables, typically labeled X and Y.
```

```python
# Line Chart - EV Adoption Trend

year_trend = df['Model Year'].value_counts().sort_index()

plt.figure(figsize=(10,5))

plt.plot(
    year_trend.index,
    year_trend.values,
    marker='o',
    linewidth=2
)

plt.title('EV Adoption Trend Over Time')
plt.xlabel('Model Year')
plt.ylabel('Number of EV Registrations')
plt.grid(True)

plt.show()
```

Interpretation of your above chart

EV registrations increase significantly in recent model years.

The upward trend indicates growing adoption of electric vehicles.

Improvements in battery technology, charging infrastructure, and government
incentives may have contributed to this growth.

The trend suggests continued expansion of the EV market.

```python
# BAR CHART : Top 10 EV Manufacturers
top_make = df['Make'].value_counts().head(10)

plt.figure(figsize=(8,5))
sns.barplot(x=top_make.values,
            y=top_make.index)
plt.title('Top 10 EV Manufacturers')
plt.show()
```

Interpretation of your above chart

Identifies market-leading manufacturers.

Shows brand dominance in EV adoption.

Useful for competitive benchmarking.

```python
# SCATTER PLOT : Electric Range vs Electric Range Log

# # This scatter plot was used to visualize the impact of log transformation applied to Electric Range and verify the reduction of skewness.

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='Range_Miles',
    y='Electric Range Log',
    alpha=0.6
)

plt.title('Electric Range vs Log Transformed Range')
plt.xlabel('Electric Range (Miles)')
plt.ylabel('Electric Range Log')
plt.show()
```

Interpretation of your above chart

Strong positive relationship between Electric Range and Electric Range Log.

Data points show a clear upward trend.

Log transformation reduces the impact of high-range outliers.

The transformed variable preserves the original pattern while making the data more balanced.



### Analysis about EV Registrations

```python
# BAR CHART

county_count = df['County'].value_counts().head(10)

plt.figure(figsize=(8,6))
sns.barplot(
    x=county_count.index,
    y=county_count.values
)

plt.title('Top 10 Counties by EV Registrations')
plt.xlabel('County')
plt.ylabel('Number of EVs')
plt.xticks(rotation=45)
plt.show()
```

```python
# BAR CHART : Average Electric Range by Manufacturer

avg_range = (
    df.groupby('Make')['Range_Miles']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)
plt.figure(figsize=(10,5))

sns.barplot(
    x=avg_range.values,
    y=avg_range.index
)

plt.title('Average EV Range by Manufacturer')
plt.xlabel('Average Range (Miles)')
plt.ylabel('Manufacturer')
plt.show()
```

## Interpretation of your above chart

Shows manufacturers producing longer-range vehicles.

Indicates technological competitiveness.

Useful for consumer preference analysis.

```python
# BAR CHART : Average Electric Range by Model Year

plt.figure(figsize=(10,6))

avg_range = df.groupby('Model Year')['Range_Miles'].mean().reset_index()

sns.barplot(
    data=avg_range,
    x='Model Year',
    y='Range_Miles',
    color='skyblue'
)

plt.title('Average Electric Range by Model Year')
plt.xlabel('Model Year')
plt.ylabel('Average Electric Range (Miles)')
plt.xticks(rotation=45)
plt.show()
```

## Interpretation of your above chart

Compares average electric range across model years.

Electric range varies between model years.

Higher ranges in some years indicate advancements in EV technology.

Useful for understanding trends in vehicle performance over time.

```python
# Correlation Heatmap

plt.figure(figsize=(8,5))

sns.heatmap(
    df.select_dtypes(include=['int64','float64']).corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

## Interpretation of your above chart

Shows relationships among numerical variables.

Strong correlations may indicate important business drivers.

Helps identify variables influencing EV characteristics.

```python
# MULTIVARIATE ANALYSIS                   # refers to a set of statistical techniques used to examine datasets containing more than two variables simultaneously.
```

```python
# Pivot Table Analysis

pivot = pd.pivot_table(
    df,
    values='Range_Miles',
    index='Make',
    columns='EV_Type',
    aggfunc='mean'
)

pivot.head()                             # o/p shows the NaN here does NOT mean missing values in the dataset.
                                         # It means: That manufacturer does not have records for that EV type in the dataset.
```

## Interpretation of your above chart

Compares average range across manufacturers and vehicle types.

Helps identify market leaders in each category.

Supports strategic product comparisons.

```python
# Heatmap from Pivot Table

plt.figure(figsize=(12,6))

sns.heatmap(
    pivot,
    annot=True,
    cmap='YlGnBu'
)

plt.title('Average Electric Range by Make and EV Type')
plt.show()
```

## Interpretation of your above chart

Easily identifies high-performing manufacturer-type combinations.

Highlights technology differences.

Useful for benchmarking EV performance.
